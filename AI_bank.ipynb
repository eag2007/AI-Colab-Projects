```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LinearRegression
from sklearn import linear_model
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import r2_score
```

Чтение и просмотр данных

```python
df = pd.read_csv("/content/Кредиты.csv", skipinitialspace=True, sep=",")
df
```

Обработаем столбец "название банка"

```python
df["название_банка_числа"] = df["название банка"].apply(lambda x: 1
                                                  if x == "ВАШ БАНК" else 0)
df
```

Проверка данных

```python
print(df.shape)
df = df.dropna()
df.shape
```

Проверка дубликатов, удаление дубликатов

```python
print(df.shape)
df = df.drop_duplicates()
df.shape
```

Разметим данные X, Y

```python
x = df[["сумма займа", "процент годовых"]]
print(x.head(10))
y = df[["название_банка_числа"]]
print(y.head(10))
```

Зададим тренировочную и тестовую выборку

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.20,
                                                    random_state=42)
x_train = x_train.values
y_train = y_train.values.ravel()
print(x_train)
print(y_train)

x_test = x_test.values
y_test = y_test.values.ravel()
```

Предскажем методом 1 логической регрессии


```python
# создание и обучение модели
model_logistic_regression = LogisticRegression()
model_logistic_regression.fit(x_train, y_train)

# предсказание значений
y_pred = model_logistic_regression.predict(x_test)

# вычисление точности

accuracy = accuracy_score(y_test, y_pred)
print(f"Точность модели: {accuracy}")

# проверка
print(round(model_logistic_regression.predict(np.array([[16600, 1]]))[0])) # 1
print(round(model_logistic_regression.predict(np.array([[2200, 8]]))[0])) # 0
```

Разметим снова данные


```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.20,
                                                    random_state=42)
x_train = x_train.values
y_train = y_train.values.ravel()

x_test = x_test.values
y_test = y_test.values.ravel()
```

Предскажем методом 2 линейной регрессии

```python
# создание и обучение модели
model_linear_regression = LinearRegression()
model_linear_regression.fit(x_train, y_train)

# предсказание значений
y_pred = model_linear_regression.predict(x_test)

# вычисление точности

accuracy = r2_score(y_test, y_pred)
print(f"Точность модели: {accuracy}")

# проверка
print(round(model_linear_regression.predict(np.array([[16600, 1]]))[0])) # 1
print(round(model_linear_regression.predict(np.array([[2200, 8]]))[0]))  # 0
```

Посмотрев на точность после Линейной регрессии и Логической регрессии можем увидеть, что точность логической регрессии намного выше, поэтому оптимальнее использовать её.
